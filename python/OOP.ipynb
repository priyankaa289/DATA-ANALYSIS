```python
# With Constructor
class students:
    print("Student Details")

    def __init__(self,name,age):
        self.name=name
        self.age=age

    def show(self):
        print("Name:- ",self.name)
        print("Age:- ",self.age)

s1=students("priyanka",21)
s2=students("pratik",19)
s1.show()
s2.show()

```

```python
class employee:
    print("Employee Details\n")

    def __init__(self,name,age,salary):
        self.name=name
        self.age=age
        self.salary=salary

    def show(self):
        print("Employee Name:-",self.name)
        print("Employee age:-",self.age)
        print("Employee salary:-",self.salary)

e1=employee("priyanka",21,35000)
e2=employee("pratik",20,40000)
e3=employee("ravi",25,30000)

e1.show()
e2.show()
e3.show()
        

```

```python
#Instance Attributes
class Mobile:
    def __init__(self ,brand, price):

        self.brand=brand
        self.price=price

m1=Mobile("Samsung",55000)
m2=Mobile("Apple",100000)

print(m1.brand,m1.price)
print(m2.brand,m2.price)
```
