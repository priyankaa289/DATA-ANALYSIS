```python
# With Constructor
class students:
    print("Student Details")

    def __init__(self,name,age):
        self.name=name
        self.age=age

    def show(self):
        print("Name:- ",self.name)
        print("Age:- ",self.age)

s1=students("priyanka",21)
s2=students("pratik",19)
s1.show()
s2.show()

```

```python
class employee:
    print("Employee Details\n")

    def __init__(self,name,age,salary):
        self.name=name
        self.age=age
        self.salary=salary

    def show(self):
        print("Employee Name:-",self.name)
        print("Employee age:-",self.age)
        print("Employee salary:-",self.salary)

e1=employee("priyanka",21,35000)
e2=employee("pratik",20,40000)
e3=employee("ravi",25,30000)

e1.show()
e2.show()
e3.show()
        
```

```python
#Instance Attributes
class Mobile:
    def __init__(self ,brand, price):

        self.brand=brand
        self.price=price

m1=Mobile("Samsung",55000)
m2=Mobile("Apple",100000)

print(m1.brand,m1.price)
print(m2.brand,m2.price)
```

```python
# Create a Car class with brand, color, and price attributes.
class car:
    def __init__(self,brand,color,price):

        self.brand=brand
        self.color=color
        self.price=price
        
c1=car("BMW M5","Black",21000000)
c2=car("Mercedes Benz","White",25000000)
c3=car("porche Gt3 rs","Dark Grey",34000000)

print(c1.brand,c1.color,c1.price)
print(c2.brand,c2.color,c2.price)
print(c3.brand,c3.color,c3.price)
```

```python
# Create a Student class with name, roll_number, and marks.

class student:
    def __init__(self,name,roll_no,marks):
        self.name=name
        self.rollno=roll_no
        self.marks=marks

s1=student("priyanka",1,89)
s2=student("veer",2,77)
s3=student("pratik",3,86)

print(s1.name,s1.rollno,s1.marks)
print(s2.name,s2.rollno,s2.marks)
print(s3.name,s3.rollno,s3.marks)
        
```

```python
# Create an Employee class with name, department, and salary.

class Employee:
    def __init__(self,name,department,salary):
        self.name=name
        self.department=department
        self.salary=salary

e1=Employee("priyanka","IT",40000)
e2=Employee("veer","sales",25000)
e3=Employee("pratik","data",20000)
e4=Employee("meera","IT",30000)

print(e1.name,e1.department,e1.salary)
print(e2.name,e2.department,e2.salary)
print(e3.name,e3.department,e3.salary)
print(e4.name,e4.department,e4.salary)
```

```python
# Create a Product class with name and price. Display the details of three products.

class product:
    def __init__(self,name,price):
        self.name=name
        self.price=price

p1=product("maybelline newyork foundation",300)
p2=product("MARS lipstick",200)
p3=product("lakme eyeliner",350)

print(p1.name,p1.price)
print(p2.name,p2.price)
print(p3.name,p3.price)
        
```

```python
# Create a BankAccount class with account_holder and balance.

class BankAccount:
    def __init__(self,account_holder,balance):
        self.account_holder=account_holder
        self.balance=balance


b1=BankAccount("priyanka",40000)
b2=BankAccount("pratik",30000)
b3=BankAccount("jiya",24000)

print(b1.account_holder,b1.balance)
print(b2.account_holder,b2.balance)
print(b3.account_holder,b3.balance)
```
