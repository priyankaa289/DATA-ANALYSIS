```python
# Single Inheritance
class car:
    def car_name(name):
     print(f"Car Name:- {name}")

class price(car):
   def car_price(price):
      print(f"car price:-{price}")

c=price
c.car_price(200000)
c.car_name("BMW")
```

```python
class car:
    def display_name():
        print("name:- BMW")

class color(car):
    def display_color():
        print("colour:- black")

b=color
b.display_name()
b.display_color()
```

```python
# Multiple Inheritance
class car:
    def display_name():
        print("name:- BMW")
        
class type():
    def display_type():
        print("type:-M5")
        
class Color(car,type):
    def display_color():
        print("color:- black")

c=Color
c.display_name()
c.display_type()
c.display_color()
```

```python
# Multilevel Inheritance

class car:
    def display_name():
        print("name:- BMW")
        
class type(car):
    def display_type():
        print("type:-M5")
        
class Color(type):
    def display_color():
        print("color:- black")

c=Color
c.display_name()
c.display_type()
c.display_color()
```

```python
# Hierarchical Inheritance

class Teachers:
    def display_tech():
        print("category of Teachers")

class maths(Teachers):
    def class_math():
        print("maths teacher:- reena sharma")

class science(Teachers):
    def class_science():
        print("science teacher:- hinal mehta")

class english(Teachers):
    def class_english():
        print("english teacher:- priya patel")

b=maths
b.display_tech()
b.class_math()

c=science
c.display_tech()
c.class_science()

d=english
d.display_tech()
d.class_english()
```

```python
class A:
    def display_a():
        print("class A")

class B(A):
    def display_b():
        print("class B")
class C(A):
    def display_c():
        print("class C")

class D(B,C):
    def display_d():
        print("class D")  

b=B
b.display_a()
b.display_b()

c=C
c.display_a()
c.display_c()

d=D
d.display_b()
d.display_c()
d.display_d()
```
